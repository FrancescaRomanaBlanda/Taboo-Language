```python
import pandas as pd
import numpy as np
```

```python

url = "/kaggle/input/datasets/shashk99/off-words/sold_test.csv"

data1 = pd.read_csv(url)
 

data1.head()
```

```python

url2 = "/kaggle/input/datasets/shashk99/off-words/sold_train.csv"

data2 = pd.read_csv(url2)

data2.head()
```

# merge data 

```python

# Merge the two datasets
merged_data = pd.concat([data1, data2], ignore_index=True)

# Remove duplicate rows
merged_data = merged_data.drop_duplicates()

# Check result
print("Total rows after merge:", len(merged_data))

# Save to CSV (optional)
merged_data.to_csv("merged_dataset.csv", index=False)
```
