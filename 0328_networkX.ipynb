```python
import pandas as pd

def extract_groups(grp):
    if pd.isna(grp):
        return []
    
    items = grp.split(" & ")
    groups = []
    
    for item in items:
        if "-" in item:
            groups.append(item.split("-")[1])
    
    return groups

df["groups"] = df["GRP"].apply(extract_groups)
```

```python
from itertools import combinations

edges = []

for groups in df["groups"]:
    if len(groups) > 1:
        edges.extend(list(combinations(groups, 2)))
```

```python
# 네트워크 만들기

import networkx as nx

from collections import Counter

edge_counts = Counter(edges)

G = nx.Graph()

for (a,b), w in edge_counts.items():
    G.add_edge(a,b, weight=w)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,10))
nx.draw(G, with_labels=True, node_size=500)
plt.show()
```

```python
weights = [G[u][v]['weight'] for u,v in G.edges()]

sizes = [G.degree(n, weight='weight') * 50 for n in G.nodes()]

plt.figure(figsize=(10,10))
nx.draw(
    G,
    with_labels=True,
    node_size=sizes,
    width=weights
)

plt.show()
```

```python
cluster1 = {
    'asian',
    'black',
    'catholic',
    'christian',
    'female',
    'islam',
    'white'
}

cluster2 = {
    'conservative',
    'feminist',
    'homosexual',
    'progressive',
    'queer'
}

cluster3 = {
    'chinese',
    'indian',
    'korean_chinese',
    'male',
    'southeast_asian'
}

cluster4 = {
    'age',
    'others',
    'socioeconomic_status'
}
```

```python
color_map = []

for node in G.nodes():
    if node in cluster1:
        color_map.append('skyblue')
    elif node in cluster2:
        color_map.append('lightgreen')
    elif node in cluster3:
        color_map.append('orange')
    elif node in cluster4:
        color_map.append('pink')
    else:
        color_map.append('grey')

sizes = [G.degree(n, weight='weight') * 50 for n in G.nodes()]
weights = [G[u][v]['weight'] for u,v in G.edges()]

plt.figure(figsize=(10,10))

pos = nx.spring_layout(G, k=0.5)

nx.draw(
    G,
    pos,
    with_labels=True,
    node_size=sizes,
    node_color=color_map,
    width=weights
)

plt.show()
```

```python
# ① weighted degree
sorted(dict(G.degree(weight="weight")).items(), key=lambda x: x[1], reverse=True)[:10]
```

```python
# ② betweenness centrality
import networkx as nx
G_dist = nx.Graph()

for u, v, data in G.edges(data=True):
    w = data["weight"]
    G_dist.add_edge(u, v, weight=w, distance=1 / w)

sorted(
    nx.betweenness_centrality(G_dist, weight="distance").items(),
    key=lambda x: x[1],
    reverse=True
)[:10]
```

→ 인종 + 종교 기반 집단 정체성 공격

2️⃣ Ideology + Feminism + Sexuality ****

3️⃣ Ethnicity cluster : East Asian / Immigrant cluster
=> 특정 국적·민족 집단을 중심으로 한 공격 cluster

4️⃣ Age + Class
→ 사회적 지위 / 계층 관련 공격

taboo는 단어가 아니라 정체성 네트워크 속에서 형성된다.

```python
# 어떤 Target pair 가 같이 등장했는가?

df['targets'] = df['GRP'].str.split(' & ')
```

```python
from itertools import combinations

pairs = []

for targets in df['targets'].dropna():
    if len(targets) >= 2:
        pairs.extend(combinations(targets, 2))
```

```python
from collections import Counter

pair_counts = Counter(pairs)
```

```python
pair_counts.most_common(20)
```

```python
import pandas as pd

pair_df = pd.DataFrame(pair_counts.items(), columns=['pair','count'])
pair_df = pair_df.sort_values(by='count', ascending=False)

pair_df.head(20)
```

```python
import matplotlib.pyplot as plt

counts = pair_df['count']

plt.hist(counts, bins=20)
plt.title("Edge Weight Distribution")
plt.xlabel("Co-target frequency")
plt.ylabel("Number of pairs")
plt.show()
```
